```python
# SMOTE (Synthetic Minority Over-sampling Technique)
#why we need Smote?
#Our dataset is highly imbalanced

# Geniune(class =0)--> 98.4%
# Fraud (class=1)---> 1.6%

# problem
# Model will be baised


#What smote does?
# Smote Generates synthetic samples for the minority class

#How smote works
```

```python
#
```

```python
from imblearn.over_sampling import SMOTE
```

```python
smote=SMOTE(random_state=42)
```

```python
x_train_smote,y_train_smote=smote.fit_resample(x_train,y_train)
```

```python
print('Before',y_train.value_counts())
print('After',y_train_smote.value_counts())
```

```python
# now data is balanced import model
```

```python
from sklearn.ensemble import RandomForestClassifier


rf=RandomForestClassifier(n_estimators=100,random_state=42)

rf.fit(x_train_smote,y_train_smote)

pred=rf.predict(x_test)
```

```python
from sklearn.metrics import *
```

```python
#accuracy_score(y_test,pred)*100
```

```python
print(classification_report(y_test,pred))
```

```python
rf_accuracy=accuracy_score(y_test, pred)
rf_precision=precision_score(y_test, pred)
rf_recall= recall_score(y_test, pred)
rf_f1=f1_score(y_test, pred)
rf_roc_auc=roc_auc_score(y_test, pred)

print("Accuracy :", rf_accuracy)
print("Precision:",rf_precision )
print("Recall   :",rf_recall )
print("F1 Score :",rf_f1 )
print("ROC-AUC  :",rf_roc_auc )
```

```python
# Confusion Matrix
rf_cm = confusion_matrix(y_test, pred)
print("Confusion Matrix :\n", rf_cm)
```

```python
#logistic model
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=1000, random_state=42)

lr.fit(x_train_smote, y_train_smote)

lr_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, lr_pred))
```

```python
comparison = pd.DataFrame({
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1 Score', 'ROC-AUC'],
    'Random Forest': [
        rf_accuracy,
        rf_precision,
        rf_recall,
        rf_f1,
        rf_roc_auc
    ],
    'Logistic Regression': [
        lr_accuracy,
        lr_precision,
        lr_recall,
        lr_f1,
        lr_roc_auc
    ]
})

comparison
```

### Model Comparison

I compared **Logistic Regression** and **Random Forest** for credit card fraud detection.

Random Forest gave **90.52% precision, 78.36% recall, and 84.00% F1-score**.

Logistic Regression gave **85.82% recall** and **97.33% ROC-AUC**, but its precision was only **11.84%** and F1-score was **20.81%**.

I selected **Random Forest** because it gave a better balance between finding fraud and reducing false alerts.

### Conclusion

**Random Forest was the better model for this project.**
